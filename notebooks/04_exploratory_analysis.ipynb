# 北上广深租房市场数据分析

## 第四部分：数据探索分析

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/processed/rent_cleaned.csv")

df = pd.read_csv(DATA_PATH)

print("数据规模：", df.shape)
```

## 一、城市租金比较

```python
# 整租和合租的面积含义不同，因此分别建立分析数据
entire_df = df[
    df["type"] == "整租"
].copy()

shared_df = df[
    df["type"] == "合租"
].copy()
```

```python
# 分别计算每个城市的整租房源数量、平均租金和租金中位数
entire_city_count = entire_df.groupby(
    "city"
)["rent_price_num"].count()

entire_city_mean = entire_df.groupby(
    "city"
)["rent_price_num"].mean().round(2)

entire_city_median = entire_df.groupby(
    "city"
)["rent_price_num"].median()
```

```python
# 把三项统计结果组合成一张表
entire_city_rent = pd.DataFrame({
    "整租房源数量": entire_city_count,
    "整租平均月租金": entire_city_mean,
    "整租月租金中位数": entire_city_median
})
```

```python
# 按照租金中位数从高到低排列
entire_city_rent = entire_city_rent.sort_values(
    "整租月租金中位数",
    ascending=False
)

entire_city_rent
```

```python
# 单独计算每个城市的合租房源数量、平均租金和租金中位数
shared_city_count = shared_df.groupby(
    "city"
)["rent_price_num"].count()

shared_city_mean = shared_df.groupby(
    "city"
)["rent_price_num"].mean().round(2)

shared_city_median = shared_df.groupby(
    "city"
)["rent_price_num"].median()
```

```python
# 把三项统计结果组合成一张表
shared_city_rent = pd.DataFrame({
    "合租房源数量": shared_city_count,
    "合租平均月租金": shared_city_mean,
    "合租月租金中位数": shared_city_median
})
```

```python
# 按照租金中位数从高到低排列
shared_city_rent = shared_city_rent.sort_values(
    "合租月租金中位数",
    ascending=False
)

shared_city_rent
# 发现北京无合租房源
```

```python
# # 统一检查每个城市的整租、合租和总房源数量
city_type_count = pd.crosstab(
    df["city"],
    df["type"],
    margins=True,
    margins_name="总计"
)

city_type_count
```

## 四、城市房源结构比较

```python
# 复制前面已经统计的城市出租类型数量表
city_rental_structure = city_type_count.copy()
```

```python
# 计算每个城市的整租占比
city_rental_structure["整租占比（%）"] = (
    city_rental_structure["整租"]
    / city_rental_structure["总计"]
    * 100
).round(2)

# 计算每个城市的合租占比
city_rental_structure["合租占比（%）"] = (
    city_rental_structure["合租"]
    / city_rental_structure["总计"]
    * 100
).round(2)

city_rental_structure
```

```python
# 统计每个城市不同卧室数量的整租房源数
entire_bedroom_count = pd.crosstab(   # pd.crosstab()：统计两个分类字段组合后出现的数量
    entire_df["city"],  # 放在行上的字段
    entire_df["bedroom_num"]  # 放在列上的字段
)

entire_bedroom_count
```

```python
# 计算每个城市内部不同卧室数量的整租房源占比
entire_bedroom_percent = pd.crosstab(
    entire_df["city"],
    entire_df["bedroom_num"],
    normalize="index"  # 表示把数量转换为比例
)

# 把小数转换为百分比，并保留两位小数
entire_bedroom_percent = (
    entire_bedroom_percent
    * 100
).round(2)

entire_bedroom_percent
```

```python
# 统计每个城市不同卧室数量的合租房源数
shared_bedroom_count = pd.crosstab(
    shared_df["city"],
    shared_df["bedroom_num"]
)

shared_bedroom_count
```

```python
# 计算每个城市内部不同卧室数量的合租房源占比
shared_bedroom_percent = pd.crosstab(
    shared_df["city"],
    shared_df["bedroom_num"],
    normalize="index"
)

# 把小数转换为百分比，并保留两位小数
shared_bedroom_percent = (
    shared_bedroom_percent
    * 100
).round(2)

shared_bedroom_percent
```

## 数据分析总结

### 数据基本情况

- 本次分析使用第3天生成的清洗数据，共11,978条住宅租房记录和26个字段。
- 数据包括11,564条整租房源和414条合租房源。
- 北京有2,992条房源，上海有2,995条，广州有2,992条，深圳有2,999条。
- 北京没有合租样本，上海只有3条合租样本，广州有94条，深圳有317条。
- 合租样本数量较少且城市分布不均衡，因此不能进行可靠的四城市合租市场排名。
- 整租与合租的租金、面积和每平方米租金均分开计算，没有混合分析。

### 主要分析结果

- 北京的整租月租金中位数最高，为6,500元；上海为6,000元，深圳为4,500元，广州为3,100元。
- 上海的整租平均月租金最高，为9,364.25元，但中位数低于北京，说明平均数受到部分高价房源影响。
- 上海的整租面积中位数最大，为90平方米；北京为78平方米，广州为73平方米，深圳为54平方米。
- 北京的整租每平方米租金中位数最高，为93.02元；深圳为85.03元，上海为80.88元，广州为47.13元。
- 上海总租金较高与其整租面积较大有关；深圳典型整租面积最小，但每平方米租金水平较高。
- 北京和上海的整租房源都以两室为主，占比分别为45.66%和40.11%。
- 广州一室、两室和三室整租占比较为接近，户型结构相对均衡。
- 深圳一室整租占47.65%，是四个城市中一室整租占比最高的城市。
- 合租记录中的`bedroom_num`含义不统一，卧室数量表只能描述当前字段的分布，不能解释为城市真实的合租户型占比。
- 合租记录中的`rent_area_num`可能混合出租房间面积和整套住宅面积，因此合租面积和每平方米租金不能用于城市市场排名。
